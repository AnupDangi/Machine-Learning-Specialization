**MultiClass Classification**
  Multiclass Classification

Definition: A classification problem with more than two possible output labels.

- 1. Recognizing digits (0-9).
- Classifying multiple diseases.
- Visual defect inspection in manufacturing (e.g., scratches, discoloration).
  
- 2.Decision Boundaries:
- Models estimate probabilities for multiple classes instead of just one.
-  Creates boundaries in feature space to separate different classes.

**MultiLabel Classification**

- Definition: Assigns multiple labels to a single input.
- Example: An image can be labeled with a car, bus, and pedestrian simultaneously.
- Output: A vector indicating the presence (1) or absence (0) of each label.
