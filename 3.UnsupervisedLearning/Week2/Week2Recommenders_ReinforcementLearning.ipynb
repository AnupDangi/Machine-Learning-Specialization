## Understanding Recommendation Systems

**Recommendation systems are widely used in online platforms like Amazon and Netflix to suggest products, movies, or services based on user preferences.
They play a crucial role in driving sales and enhancing user experience by predicting what users might like based on their past ratings.**

- Framework of a Recommendation System

The system consists of users and items (e.g., movies), where users rate items on a scale (e.g., 1 to 5 stars).
The framework includes variables to denote users, items, and their ratings, allowing the system to identify which items have not been rated by users.
Predicting User Ratings

The goal is to predict how users would rate unrated items, enabling the system to recommend items that users are likely to rate highly.
The upcoming content will explore algorithms for making these predictions, starting with the assumption that additional features about the items are available.

**Understanding Feature Derivation**

- In collaborative filtering, if you lack predefined features (like romance or action), you can infer them from user ratings and parameters.
By analyzing user ratings and their associated parameters, you can make educated guesses about the features of each movie.
Cost Function for Learning Features

- A cost function is established to minimize the squared error between predicted ratings and actual ratings, allowing for the learning of features for each movie.
Regularization terms can be added to prevent overfitting, ensuring that the model generalizes well.
Collaborative Filtering Algorithm

- The overall cost function combines the learning of user parameters and movie features, allowing for simultaneous optimization.
By minimizing this cost function using techniques like gradient descent, you can effectively learn the parameters for users and features for movies, enhancing the recommendation system.

**Understanding Mean Normalization**

- Mean normalization helps algorithms run more efficiently by adjusting movie ratings to have a consistent average value, which is particularly beneficial for new users who haven't rated any movies.
- By calculating the average rating for each movie and adjusting the ratings accordingly, the algorithm can make more reasonable predictions for users like Eve, who have no prior ratings.
Impact on Predictions

- Without mean normalization, new users' ratings would default to zero, leading to unhelpful predictions. Mean normalization allows the algorithm to predict ratings based on the average ratings of other users.
- This adjustment results in more realistic predictions, as it uses the average ratings rather than assuming a new user will rate everything as zero.

- Implementation Considerations

Normalizing the rows of the rating matrix is emphasized as more important than normalizing the columns, especially for new users with few ratings.
The practice of mean normalization not only enhances prediction accuracy but also speeds up the optimization process for the recommendation system.

## Understanding Collaborative vs. Content-Based Filtering

**Collaborative filtering recommends items based on ratings from similar users, relying on user interactions with items.
Content-based filtering recommends items based on the features of both users and items, allowing for more personalized recommendations.
Features of Users and Items**

**User features can include demographics (age, gender, country) and past behaviors (ratings and genres of movies watched).
Item features can include attributes like genre, year of release, and average ratings, which help in matching users with suitable items.
Vector Representation and Matching**

**In content-based filtering, user and item features are represented as vectors (vj_u for users and vi_m for items) to facilitate the prediction of ratings through dot products.
The challenge lies in computing these vectors to accurately represent user preferences and item characteristics.**

## Developing a content-based filtering algorithm using deep learning, highlighting the architecture of user and movie networks and how they interact to predict ratings.

- User and Movie Networks: User features (age, gender) create a user vector ( v_u ), while movie features (year of release) create a movie vector ( v_m ). Both networks can differ in architecture but must have the same output size for predictions.

- Training the Model: A cost function minimizes the squared error between predicted and actual ratings, adjusting both networks' parameters. Regularization can help prevent overfitting.

- Finding Similar Items: The model can identify similar movies by calculating vector distances, and pre-computation of similar items enhances recommendation speed.

## Recommender systems efficiently recommend items from large catalogs. The process involves two main steps:

**Retrieval Step:** 
- Generates a broad list of potential items based on user preferences and previously watched items.
- It may include similar items and popular items in genres the user enjoys.

**Ranking Step:**
- Refines the list by predicting ratings for the retrieved items using a neural network.
- The best items are then displayed to the user based on predicted ratings.
- This two-step approach allows for both speed and accuracy in recommendations, even with large datasets. Ethical considerations in building these systems are also highlighted.

 ## Recommender systems   impact on society. 
 
 **Key points include:**

- Use Cases: Recommender systems can enhance user experience by suggesting relevant content, but they can also lead to negative outcomes, such as promoting harmful products or content.
- Profit Maximization: Companies often prioritize profit over user satisfaction, which can result in recommending products that are more profitable rather than those that are most beneficial to users.
- Transparency: There is a need for transparency in how recommendations are made, to build trust with users.
- Potential Harm: Some systems may amplify harmful content, like conspiracy theories, due to their engaging nature.
- Ameliorations: Suggestions include filtering out exploitative businesses and being transparent about recommendation criteria.
