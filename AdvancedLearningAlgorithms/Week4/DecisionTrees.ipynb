## Decision Tree Learning

**Entropy quantifies the impurity of a dataset, with values ranging from 0 (pure) to 1 (most impure). A 50-50 mix of two classes (e.g., cats and dogs) results in the highest entropy.
For example, in a set of six examples with three cats and three dogs, the entropy is 1, indicating maximum impurity.**

**One-hot encoding**
- One-Hot Encoding: A technique to convert categorical features with multiple values into binary features (0 or 1).

- Example: Instead of a single feature for ear shape (pointy, floppy, oval), create three binary features:

- Pointy ears
- Floppy ears
- Oval ears

- Benefits:
 - Simplifies decision tree learning by ensuring each feature only takes on two values.
- Applicable to various machine learning models, including neural networks and logistic regression.
- Application: Practice using one-hot encoding on different datasets to reinforce understanding.

**Regression Algorithm**

- Decision Trees: Used for both classification and regression problems.
- Regression Trees: Predict a numerical value (e.g., weight of an animal) instead of classifying categories.
- Splitting Criteria: In regression, the goal is to reduce variance rather than entropy.
- Variance Calculation: Variance measures how much a set of numbers varies; lower variance indicates better predictions.
- Feature Selection: Choose the feature that provides the largest reduction in variance for splits.
- Ensemble Methods: Training multiple decision trees can yield better results than a single tree.

**Tree Ensembles**
- A tree ensemble is a collection of multiple decision trees that work together to make predictions.
Instead of relying on a single decision tree, which can be sensitive to small changes in data, a tree ensemble combines the predictions from several trees.


<img src="images/treeensemble.png" width="500px" height="500px"/>


<img src="images/difftreeensemble.png" width="500px" height="500px"/>


**Benefits of using tree ensemble:**
- Robustness: Reduces the impact of individual tree errors by averaging predictions.
- Diversity: Each tree learns different patterns, enhancing overall understanding.
- Voting Mechanism: Trees vote on predictions; the majority vote determines the final outcome.
- Reduced Overfitting: Combines multiple trees to avoid capturing noise from training data.
- Improved Accuracy: Generally leads to better prediction accuracy compared to a single decision tree

**Sampling with Replacement**
- Sampling with Replacement: A technique used to create multiple random training sets from an original dataset.
- Demonstration: Illustrated using colored tokens (red, yellow, green, blue) to show how tokens are picked, replaced, and sampled multiple times.
- Importance: Ensures diversity in training sets by allowing the same example to be selected multiple times, leading to slightly different datasets.
- Application: Critical for building an ensemble of trees in machine learning, where new training sets are created that are similar yet different from the original.

**Random Forest Algorithm**
- Random Forest is an ensemble learning algorithm that constructs multiple decision trees during training and outputs the mode of their predictions (for classification) or the mean prediction (for regression). It enhances prediction accuracy and robustness by combining the results of various trees, which are generated using different subsets of the training data and features.

- Ensemble Method: Combines multiple decision trees.
- Sampling with Replacement: Creates diverse training sets from the original dataset.
- Feature Randomization: Selects a random subset of features for each tree to improve diversity.
- Robustness: Averages predictions to reduce the impact of outliers and improve accuracy.
